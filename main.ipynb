```python
import matplotlib.pyplot as plt
# Sample data
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]
# Plot the data
plt.plot(x, y, color='blue', marker='o', linestyle='--')
# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Graph')
# Show the graph
plt.show()
```
